```python
import gradio as gr

def charger_fichier(fichier_à_charger):
    with open(fichier_à_charger, 'r', encoding='utf-8') as fichier:
        contenu = fichier.read()
        return contenu
    

def nettoyer_fichier(fichier_à_nettoyer):
    a = charger_fichier(fichier_à_nettoyer)
    contenu_nettoyé = a.encode('ascii', 'ignore').decode('ascii')
    with open('atterissage.json', 'w') as fichier1:
        fichier1.write(contenu_nettoyé)
        return contenu_nettoyé
    
iface2 = gr.Interface(fn=nettoyer_fichier, inputs=gr.Textbox( placeholder="Entrez le nom du fichier ici..."),outputs=gr.Textbox( lines=6955))
iface2.launch()
```
